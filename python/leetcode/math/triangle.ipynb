```python
triangle = [[-1],[-2,-3]]
```

```python
S = len(triangle)
total = 0
for i in range(S):  # 1
    min = triangle[i][0] # min = -2
    for j in range(i): # 0 - 1
        if min > triangle[i][j]:
            min = triangle[i][j]
    total += min

total
```
